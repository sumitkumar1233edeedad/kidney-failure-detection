```python
import pandas as pd
from scipy.io import arff

# Load the .arff file
data, meta = arff.loadarff('chronic_kidney_disease.arff')

# Convert to DataFrame
df = pd.DataFrame(data)

# Decode byte strings (ARFF files encode strings as bytes)
df = df.apply(lambda col: col.str.decode('utf-8')
              if col.dtype == object else col)

# Check the data
print(df.shape)
print(df.head())
print(df.columns.tolist())
print(df.dtypes)
```
